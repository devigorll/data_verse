Seja bem-vindo(a) à oitava aula! Hoje vamos aprender a automatizar tarefas repetitivas. Você aprenderá como utilizar o loop `for` para repetições contadas e iterações, o loop `while` para repetições baseadas em condições, e os comandos de controle `break` e `continue`.

### 1. O Loop `for` e a Função `range()`

O loop `for` é utilizado quando sabemos previamente quantas vezes um bloco de código deve ser executado ou quando queremos percorrer (iterar) os itens de uma sequência (como listas, tuplas ou textos).

* `range(stop)`: Gera uma sequência de 0 até `stop - 1`.
* `range(start, stop, step)`: Gera uma sequência com início, fim e passo customizados.

```python
# Repetindo um bloco 5 vezes com range()
print("--- Contagem Simples ---")
for i in range(5):
    print(f"Número: {i}")

# Percorrendo uma lista diretamente
print("\n--- Iterando sobre Lista ---")
frutas = ["Maçã", "Banana", "Laranja", "Uva"]
for fruta in frutas:
    print(f"Fruta: {fruta}")
```

---

### 2. O Loop `while` e Comandos de Controle

O loop `while` executa um bloco de código **enquanto** uma condição específica for verdadeira (`True`).

⚠️ **Cuidado com o Loop Infinito:** Garanta sempre que a condição do `while` mude em algum momento para evitar que o programa trave!

#### Comandos de Controle:
* `break`: Interrompe e encerra o loop imediatamente.
* `continue`: Pula a iteração atual e vai para a próxima ciclo do loop.

```python
# Exemplo de loop while
contador = 1
print("--- Loop While ---")
while contador <= 5:
    print(f"Contador em: {contador}")
    contador += 1  # Incremento fundamental para evitar loop infinito

# Usando break e continue
print("\n--- Break e Continue ---")
for numero in range(1, 10):
    if numero == 3:
        continue  # Pula o número 3
    if numero == 7:
        break     # Para a execução ao chegar no 7
    print(f"Número processado: {numero}")
```

### 🚀 Desafio da Aula 08: Validador de Senha com Tentativas

Crie um script em Python que:
1. Defina uma senha correta padrão (ex: `"python123"`).
2. Permita ao usuário tentar adivinhar a senha no máximo **3 vezes** usando um loop `while` ou `for`.
3. Se o usuário acertar a senha, exiba uma mensagem de boas-vindas e encerre a repetição imediatamente com `break`.
4. Se o usuário errar em todas as 3 tentativas, informe que o acesso foi bloqueado.

```python
# Resolução do Desafio
SENHA_CORRETA = "python123"
tentativas_maximas = 3
tentativa_atual = 1
acertou = False

while tentativa_atual <= tentativas_maximas:
    senha_digitada = input(f"Tentativa {tentativa_atual}/{tentativas_maximas} - Digite a senha: ")
    
    if senha_digitada == SENHA_CORRETA:
        print("\n✅ Acesso concedido! Seja bem-vindo ao DataVerse.")
        acertou = True
        break
    else:
        print("❌ Senha incorreta!")
        tentativa_atual += 1

if not acertou:
    print("\n🔒 Você excedeu o número máximo de tentativas. Acesso Bloqueado!")
```
