### 1. O que são Funções e a Palavra-chave `def`

Funções são blocos de código reutilizáveis que realizam uma tarefa específica. Em Python, definimos uma função usando a palavra-chave `def`.

#### Conceitos Fundamentais:
* **Parâmetros:** Variáveis declaradas na definição da função para receber dados externos.
* **Argumentos:** Os valores reais passados para a função quando ela é chamada.
* **`print` vs `return`:** O `print()` apenas exibe uma mensagem no terminal, enquanto o `return` **devolve** um valor para quem chamou a função, permitindo que esse valor seja guardado em uma variável ou reutilizado em outros cálculos.

```python
# Definindo uma função simples sem retorno
def saudar(nome):
    print(f"Olá, {nome}! Bem-vindo à aula de funções do DataVerse.")

# Chamando a função
saudar("Igor")
saudar("Maria")
```

---

### 2. Parâmetros Padrão, Retorno e Escopo

* **Parâmetros Padrão (Opcionais):** Permitem definir um valor *default* caso o argumento não seja informado na chamada.
* **Retorno de Valores (`return`):** Encerra a execução da função e devolve o resultado.
* **Escopo de Variáveis:** Variáveis criadas dentro de uma função são **locais** (só existem ali dentro), enquanto variáveis criadas fora são **globais**.

```python
# Função com retorno e parâmetro padrão
def calcular_desconto(preco, percentual=10):
    desconto = preco * (percentual / 100)
    preco_final = preco - desconto
    return preco_final

# Usando o parâmetro padrão (10% de desconto)
valor_com_10_porcento = calcular_desconto(100.0)
print(f"Preço com 10% de desconto: R$ {valor_com_10_porcento:.2f}")

# Passando um percentual customizado (20% de desconto)
valor_com_20_porcento = calcular_desconto(100.0, 20)
print(f"Preço com 20% de desconto: R$ {valor_com_20_porcento:.2f}")
```

### 🚀 Desafio da Aula 09: Conversor de Temperatura Reutilizável

Crie um script em Python que:
1. Defina uma função chamada `celsius_para_fahrenheit(celsius)` que receba uma temperatura em Celsius e **retorne** a temperatura equivalente em Fahrenheit.
   * *Fórmula:* $F = (C \times 1.8) + 32$
2. Defina uma função chamada `celsius_para_kelvin(celsius)` que receba uma temperatura em Celsius e **retorne** a temperatura em Kelvin.
   * *Fórmula:* $K = C + 273.15$
3. Solicite ao usuário uma temperatura em Celsius, chame ambas as funções e exiba os resultados formatados com 2 casas decimais.

```python
# Resolução do Desafio

# 1. Função para converter Celsius para Fahrenheit
def celsius_para_fahrenheit(celsius):
    return (celsius * 1.8) + 32

# 2. Função para converter Celsius para Kelvin
def celsius_para_kelvin(celsius):
    return celsius + 273.15

# 3. Interação com o usuário e chamada das funções
temp_celsius = float(input("Digite a temperatura em °C: "))

temp_fahrenheit = celsius_para_fahrenheit(temp_celsius)
temp_kelvin = celsius_para_kelvin(temp_celsius)

print(f"\n🌡️ Temperatura de Entrada: {temp_celsius:.2f} °C")
print(f"👉 Em Fahrenheit: {temp_fahrenheit:.2f} °F")
print(f"👉 Em Kelvin: {temp_kelvin:.2f} K")
```
