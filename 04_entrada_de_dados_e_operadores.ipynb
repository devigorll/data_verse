### 1. Entrada de Dados com `input()`

Para permitir que o usuário interaja com nosso programa, utilizamos a função `input()`.

⚠️ **Regra de Ouro:** A função `input()` **sempre** retorna o valor digitado como texto (`string`), independentemente do que for digitado (mesmo se forem números). Se precisarmos realizar cálculos com esse valor, devemos aplicar a conversão de tipo (*casting*).

```python
# Capturando texto do usuário
nome = input("Digite o seu nome: ")
print(f"Olá, {nome}! Seja bem-vindo ao DataVerse.")

# Capturando e convertendo números (Casting)
idade = int(input("Digite a sua idade: "))
print(f"Ano que vem você terá {idade + 1} anos.")
print(f"Tipo da variável 'idade': {type(idade)}")
```

---

### 2. Operadores Aritméticos

O Python oferece diversos operadores para a realização de operações matemáticas:

| Operador | Operação | Exemplo | Resultado |
| :---: | :---: | :---: | :---: |
| `+` | Adição | `5 + 2` | `7` |
| `-` | Subtração | `5 - 2` | `3` |
| `*` | Multiplicação | `5 * 2` | `10` |
| `/` | Divisão (float) | `5 / 2` | `2.5` |
| `//` | Divisão Inteira | `5 // 2` | `2` |
| `%` | Módulo (Resto da divisão) | `5 % 2` | `1` |
| `**` | Potenciação | `5 ** 2` | `25` |

---

### 3. Precedência dos Operadores

Assim como na matemática tradicional, o Python segue uma ordem de prioridade para resolver as expressões:

1. **Parênteses** `()`
2. **Potenciação** `**`
3. **Multiplicação, Divisão, Divisão Inteira e Módulo** `*`, `/`, `//`, `%` *(da esquerda para a direita)*
4. **Adição e Subtração** `+`, `-` *(da esquerda para a direita)*

```python
# Exemplos práticos de operadores
num1 = 10
num2 = 3

print(f"Soma: {num1 + num2}")
print(f"Divisão Tradicional: {num1 / num2}")
print(f"Divisão Inteira (descarta os decimais): {num1 // num2}")
print(f"Resto da Divisão (Módulo): {num1 % num2}")
print(f"Potenciação (10 ao cubo): {num1 ** num2}")

# Exemplo de precedência
resultado_sem_parenteses = 5 + 3 * 2  # 3 * 2 = 6 -> 5 + 6 = 11
resultado_com_parenteses = (5 + 3) * 2  # 5 + 3 = 8 -> 8 * 2 = 16

print(f"Sem parênteses: {resultado_sem_parenteses}")
print(f"Com parênteses: {resultado_com_parenteses}")
```

### 🚀 Desafio da Aula 03: Calculadora de Média Escolar

Crie um script em Python que:
1. Peça ao usuário o nome do aluno.
2. Solicite 3 notas do aluno (podem conter números decimais/floats).
3. Calcule a média aritmética simples dessas 3 notas.
4. Imprima uma mensagem amigável exibindo o nome do aluno e a média final formatada com duas casas decimais (dica: use `:.2f`).

```python
# Resolução do Desafio
nome_aluno = input("Digite o nome do aluno: ")

nota1 = float(input("Digite a primeira nota: "))
nota2 = float(input("Digite a segunda nota: "))
nota3 = float(input("Digite a terceira nota: "))

# Cálculo da média utilizando parênteses para garantir a prioridade da soma
media = (nota1 + nota2 + nota3) / 3

print(f"A média final do aluno {nome_aluno} foi: {media:.2f}")
```
