### 1. O que são Strings e Imutabilidade

Uma **string** é uma sequência de caracteres encadeados. No Python, textos são imutáveis (uma vez criados, não podem ser alterados diretamente, apenas transformados em novas strings).

Cada caractere em uma string possui uma posição de índice, começando sempre do **0**.

```python
# Interpolação Moderna com f-strings
nome = "Igor"
canal = "DataVerse"

# f-strings facilitam a inclusão de variáveis dentro do texto
mensagem = f"Olá! Meu nome é {nome} e você está no canal {canal}."
print(mensagem)
```

---

### 2. Métodos Fundamentais de Strings

O Python já vem com vários métodos prontos para manipular textos. Veja os principais:

* `.upper()`: Converte todos os caracteres para MAIÚSCULAS.
* `.lower()`: Converte todos os caracteres para minúsculas.
* `.strip()`: Remove espaços em branco extras no início e no final.
* `.replace("antigo", "novo")`: Substitui um trecho do texto por outro.
* `.split("delimitador")`: Divide o texto em uma lista com base em um separador.

```python
texto = "   DataVerse - Ciência de Dados   "

# Removendo espaços sobressalentes
texto_limpo = texto.strip()
print(f"Texto sem espaços: '{texto_limpo}'")

# Alterando caixa de texto
print(f"Maiúsculas: {texto_limpo.upper()}")
print(f"Minúsculas: {texto_limpo.lower()}")

# Substituindo caracteres
texto_modificado = texto_limpo.replace("Ciência", "Análise")
print(f"Texto substituído: {texto_modificado}")

# Divisão de texto (separando por traço)
palavras = texto_limpo.split("-")
print(f"Texto dividido em lista: {palavras}")
```

---

### 3. Fatiamento de Strings (Slicing)

Podemos extrair partes específicas de um texto usando a sintaxe `[início:fim:passo]`.

⚠️ **Lembre-se:** O índice final é **exclusivo** (o Python pega até a posição anterior a ele).

```python
linguagem = "Python3"

# Pegando do índice 0 ao 5 (os 6 primeiros caracteres)
print(f"Primeiros 6 caracteres: {linguagem[0:6]}")

# Atalho: omitir o início significa 'começar do zero'
print(f"Atalho primeiros 6 caracteres: {linguagem[:6]}")

# Pegando os últimos caracteres com índices negativos
print(f"Último caractere: {linguagem[-1]}")
```

### 🚀 Desafio da Aula 04: Formatador de Nome do Usuário

Crie um script em Python que:
1. Peça para o usuário digitar o nome completo com espaços extras ou letras misturadas (ex: `   iGoR cRuZ   `).
2. Remova todos os espaços desnecessários do início e do fim.
3. Imprima o nome formatado todo em maiúsculas.
4. Imprima o primeiro nome do usuário usando o fatiamento ou o método `.split()`.

```python
# Resolução do Desafio
nome_completo = input("Digite seu nome completo: ")

# 1. Removendo espaços extras das pontas
nome_limpo = nome_completo.strip()

# 2. Convertendo tudo para maiúsculas
nome_maiusculo = nome_limpo.upper()
print(f"Nome formatado: {nome_maiusculo}")

# 3. Extraindo o primeiro nome
primeiro_nome = nome_limpo.split()[0]
print(f"Primeiro nome: {primeiro_nome.capitalize()}")
```
