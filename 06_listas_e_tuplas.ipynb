## 📚 Aula 06: Listas e Tuplas

Seja bem-vindo(a) à sexta aula! Hoje vamos aprender a trabalhar com estruturas capazes de armazenar múltiplos dados em uma única variável: as **Listas** e as **Tuplas**. Vamos entender as diferenças entre elas, como acessar seus elementos e os conceitos de mutabilidade.

---

### 1. Listas (`list`)

Listas são coleções ordenadas, flexíveis e **mutáveis** (podemos alterar, adicionar e remover elementos a qualquer momento). Elas são representadas por colchetes `[]`.

#### Acesso por Índices:
* **Índices Positivos:** Começam do `0` (primeiro elemento).
* **Índices Negativos:** Começam do `-1` (último elemento).

```python
# Criando uma lista de linguagens
tecnologias = ["Python", "SQL", "Power BI", "Tableau"]

# Acessando elementos por índice
print(f"Primeiro elemento: {tecnologias[0]}")
print(f"Último elemento (índice negativo): {tecnologias[-1]}")

# Alterando um elemento existente (Mutabilidade)
tecnologias[3] = "Excel"
print(f"Lista após alteração: {tecnologias}")
```

---

### 2. Métodos de Listas e a Função `len()`

O Python possui diversas funções nativas para manipular listas:

* `len(lista)`: Retorna o tamanho (quantidade de itens) da lista.
* `.append(item)`: Adiciona um novo elemento ao **final** da lista.
* `.insert(posicao, item)`: Insere um elemento em uma **posição específica**.
* `.pop(indice)`: Remove e retorna o item do índice indicado (se omitido, remove o último).
* `.remove(valor)`: Remove a **primeira ocorrência** de um valor específico.

```python
# Manipulando a lista de tecnologias
print(f"Quantidade total de itens: {len(tecnologias)}")

# Adicionando elementos
tecnologias.append("Git")
print(f"Após .append('Git'): {tecnologias}")

tecnologias.insert(1, "Pandas")
print(f"Após .insert(1, 'Pandas'): {tecnologias}")

# Removendo elementos
removido = tecnologias.pop(2) # Remove o item do índice 2
print(f"Item removido com .pop(): {removido}")

tecnologias.remove("Excel")
print(f"Após .remove('Excel'): {tecnologias}")
```

---

### 3. Tuplas (`tuple`)

Tuplas são coleções ordenadas, porém **imutáveis** (não podem ser alteradas depois de criadas). São representadas por parênteses `()`.

#### Por que usar Tuplas?
* **Segurança de Dados:** Para garantir que valores críticos (ex: coordenadas GPS, configurações, meses do ano) não sejam alterados por engano.
* **Performance:** São levemente mais rápidas e consomem menos memória que as listas.

```python
# Criando uma tupla
coordenadas_sp = (-23.5505, -46.6333)

print(f"Latitude: {coordenadas_sp[0]}")
print(f"Longitude: {coordenadas_sp[1]}")

# Tentativa de alteração gera erro no Python:
# coordenadas_sp[0] = -23.0000  # Descomentar essa linha causará um TypeError!
```

### 🚀 Desafio da Aula 06: Gerenciador de Lista de Compras

Crie um script em Python que:
1. Inicie com uma lista de compras vazia chamada `carrinho`.
2. Peça ao usuário para digitar 3 itens para adicionar ao carrinho.
3. Exiba a lista completa de compras e o total de itens salvos.
4. Pergunta ao usuário se ele deseja remover o último item adicionado.
5. Caso a resposta seja "sim" (ou "s"), remova o último item e exiba a lista finalizada.

```python
# Resolução do Desafio
carrinho = []

# 1. Adicionando 3 itens
item1 = input("Digite o 1º item para a lista de compras: ")
carrinho.append(item1)

item2 = input("Digite o 2º item para a lista de compras: ")
carrinho.append(item2)

item3 = input("Digite o 3º item para a lista de compras: ")
carrinho.append(item3)

print(f"\n🛒 Lista de Compras atual ({len(carrinho)} itens): {carrinho}")

# 2. Pergunta de remoção do último item
resposta = input("\nDeseja remover o último item adicionado? (s/n): ").strip().lower()

if resposta == 's' or resposta == 'sim':
    item_removido = carrinho.pop()
    print(f"❌ O item '{item_removido}' foi removido com sucesso!")

print(f"\n📋 Lista de Compras Final: {carrinho}")
```
