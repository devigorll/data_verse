### 1. Lógica Booleana e Comparadores

Para tomar decisões, o Python compara valores e retorna um resultado booleano: `True` (Verdadeiro) ou `False` (Falso).

#### Operadores de Comparação:
* `==` : Igual a
* `!=` : Diferente de
* `>` : Maior que
* `<` : Menor que
* `>=` : Maior ou igual a
* `<=` : Menor ou igual a

---

### 2. A Regra de Ouro da Indentação 📐

Diferente de outras linguagens que usam chaves `{}` para definir blocos de código, o Python usa a **indentação** (alinhamento do texto com 4 espaços ou TAB). Tudo o que estiver recuado à direita pertence ao bloco da condição.

```python
# Testando operadores de comparação
idade = 18

print(f"É maior ou igual a 18? {idade >= 18}")
print(f"É igual a 20? {idade == 20}")
print(f"É diferente de 15? {idade != 15}")
```

---

### 3. As Estruturas `if`, `elif` e `else`

* **`if` (se):** Executa o bloco de código apenas se a condição for verdadeira.
* **`elif` (senão se):** Avalia uma nova condição caso a anterior seja falsa.
* **`else` (senão):** Executa caso nenhuma das condições anteriores seja atendida.

#### Operadores Lógicos:
* `and` : Retorna `True` se **todas** as condições forem verdadeiras.
* `or`  : Retorna `True` se **pelo menos uma** condição for verdadeira.
* `not` : Inverte o resultado booleano (`True` vira `False` e vice-versa).

```python
# Exemplo prático: Verificador de Idade e Categoria
idade_usuario = int(input("Digite a sua idade: "))

if idade_usuario < 12:
    print("Categoria: Criança 🎈")
elif idade_usuario >= 12 and idade_usuario < 18:
    print("Categoria: Adolescente 🎧")
elif idade_usuario >= 18 and idade_usuario < 60:
    print("Categoria: Adulto 💼")
else:
    print("Categoria: Idoso 👴")
```

### 🚀 Desafio da Aula 05: Sistema de Análise de Empréstimo

Crie um script em Python para um sistema bancário que avalie a aprovação de um empréstimo:
1. Peça ao usuário o valor da sua **renda mensal**.
2. Peça o valor da **parcela desejada**.
3. Peça a **pontuação de crédito (score)** do usuário (de 0 a 1000).
4. Regra de Aprovação:
   * O empréstimo será **Aprovado** apenas se a parcela for menor ou igual a 30% da renda mensal **E** se o score for maior ou igual a 600.
   * Caso contrário, exiba que o empréstimo foi **Negado**.

```python
# Resolução do Desafio
renda = float(input("Digite a sua renda mensal (R$): "))
parcela = float(input("Digite o valor da parcela desejada (R$): "))
score = int(input("Digite o seu score de crédito (0 a 1000): "))

# Calculando o limite máximo permitido para a parcela (30% da renda)
limite_parcela = renda * 0.30

# Verificando as condições com operadores lógicos
if parcela <= limite_parcela and score >= 600:
    print("\n✅ Empréstimo APROVADO! Parabéns.")
else:
    print("\n❌ Empréstimo NEGADO.")
    if parcela > limite_parcela:
        print(f"- A parcela excede o limite de 30% da renda (Máximo: R$ {limite_parcela:.2f})")
    if score < 600:
        print("- O score de crédito está abaixo do mínimo exigido (600 pontos)")
```
