### 1. O que são Variáveis? 📦

Imagine que a memória do seu computador é um grande armário cheio de gavetas. Uma **variável** nada mais é do que uma **caixa etiquetada** onde você guarda um valor para usar mais tarde no seu código.

#### Regras e Convenções de Nomeação (PEP 8):
* Utilize o padrão **`snake_case`**: palavras em minúsculas separadas por sublinhado `_` (ex: `nome_usuario`, `idade_aluno`).
* O nome não pode começar com números nem conter acentos ou caracteres especiais.
* O Python é *case-sensitive* (diferencia maiúsculas de minúsculas: `idade` é diferente de `Idade`).

```python
# Criando nossas primeiras variáveis
nome_aluno = "Igor Cruz"
idade = 25
altura = 1.75
matriculado = True

# Exibindo os valores salvos
print(nome_aluno)
print(idade)
print(altura)
print(matriculado)
```

---

### 2. Tipos Primitivos de Dados

O Python possui quatro tipos básicos fundamentais de dados:

* **`int` (Inteiro):** Números inteiros, sem casas decimais (ex: `10`, `-5`, `0`).
* **`float` (Ponto Flutuante):** Números decimais ou fracionários, separados por ponto `.` (ex: `10.5`, `-3.14`).
* **`str` (String):** Textos delimitados por aspas simples `''` ou duplas `""` (ex: `"DataVerse"`).
* **`bool` (Booleano):** Valores lógicos: apenas `True` (Verdadeiro) ou `False` (Falso).

#### Checando o Tipo com `type()`
Podemos usar a função nativa `type()` para verificar qual é o tipo de dado armazenado em uma variável.

```python
# Verificando os tipos das variáveis com type()
print(f"Tipo da variável 'nome_aluno': {type(nome_aluno)}")
print(f"Tipo da variável 'idade': {type(idade)}")
print(f"Tipo da variável 'altura': {type(altura)}")
print(f"Tipo da variável 'matriculado': {type(matriculado)}")
```

---

### 3. Conversão de Tipos (Casting)

Às vezes precisamos converter um dado de um tipo para outro. Para isso, utilizamos as funções de conversão: `int()`, `float()`, `str()` e `bool()`.

```python
# Exemplo de Conversão (Casting)
numero_texto = "100"
print(f"Tipo original: {type(numero_texto)}")

# Convertendo de String para Inteiro
numero_inteiro = int(numero_texto)
print(f"Após int(): {type(numero_inteiro)} | Resultado + 5 = {numero_inteiro + 5}")

# Convertendo Inteiro para Float
numero_float = float(numero_inteiro)
print(f"Após float(): {numero_float} ({type(numero_float)})")
```
