## 📚 Aula 07: Dicionários e Conjuntos

Seja bem-vindo(a) à sétima aula! Hoje vamos explorar duas estruturas de dados fundamentais no Python para organizar informações complexas e tratar duplicatas: os **Dicionários** (estrutura Chave-Valor) e os **Conjuntos** (*Sets*).

---

### 1. Dicionários (`dict`)

Dicionários são coleções ordenadas (a partir do Python 3.7) e mutáveis que armazenam dados no formato de **Chave-Valor** (muito parecido com objetos JSON utilizados em APIs e web). São representados por chaves `{}`.

* **Chave (Key):** Funciona como o identificador único (como um índice nomeado).
* **Valor (Value):** O dado associado àquela chave.

```python
# Criando um dicionário de aluno
aluno = {
    "nome": "Igor Cruz",
    "curso": "Ciência de Dados",
    "semestre": 3,
    "ativo": True
}

# Lendo valores através das chaves
print(f"Nome do Aluno: {aluno['nome']}")
print(f"Curso: {aluno['curso']}")

# Atualizando e adicionando novas chaves
aluno["semestre"] = 4  # Atualiza valor existente
aluno["nota_final"] = 9.5  # Adiciona nova chave-valor

print(f"Dicionário atualizado: {aluno}")
```

---

### 2. Métodos Úteis de Dicionários

Para navegar facilmente pelas chaves e valores de um dicionário, utilizamos os seguintes métodos:

* `.keys()`: Retorna uma lista com todas as **chaves**.
* `.values()`: Retorna uma lista com todos os **valores**.
* `.items()`: Retorna os pares **(chave, valor)** em formato de tuplas.
* `.pop("chave")`: Remove a chave informada e retorna o seu valor.

```python
# Explorando os métodos de dicionário
print(f"Chaves do dicionário: {list(aluno.keys())}")
print(f"Valores do dicionário: {list(aluno.values())}")
print(f"Pares Chave-Valor: {list(aluno.items())}")

# Removendo uma chave
status = aluno.pop("ativo")
print(f"Status removido: {status}")
print(f"Dicionário final: {aluno}")
```

---

### 3. Conjuntos (`set`)

Conjuntos são coleções **não ordenadas** que **não permitem elementos duplicados**. São extremamente úteis na análise de dados para filtrar valores únicos e realizar operações matemáticas (como união e intersecção).

São criados usando `{}` (sem chave-valor) ou a função `set()`.

```python
# Lista com nomes repetidos
lista_acessos = ["Igor", "Maria", "Igor", "João", "Maria", "Ana"]

# Convertendo para set para remover duplicatas automaticamente
usuarios_unicos = set(lista_acessos)

print(f"Lista original (com duplicatas): {lista_acessos}")
print(f"Conjunto filtrado (apenas únicos): {usuarios_unicos}")

# Adicionando e removendo elementos em um Set
usuarios_unicos.add("Carlos")
usuarios_unicos.remove("João")
print(f"Set após modificações: {usuarios_unicos}")
```

### 🚀 Desafio da Aula 07: Cadastro de Produto e Estoque

Crie um script em Python que:
1. Crie um dicionário para cadastrar um produto contendo: `nome`, `preco` e `categoria`.
2. Peça para o usuário digitar esses 3 dados via `input()`.
3. Receba também uma lista de palavras-chave/tags do produto separadas por vírgula (ex: `tech, novo, tech, oferta`).
4. Converta a string de tags em um **Set** para garantir que não existam tags duplicadas.
5. Adicione esse Set de tags como uma nova chave no dicionário do produto e imprima o resultado final formatado.

```python
# Resolução do Desafio
# 1 e 2. Recebendo dados do produto
nome_prod = input("Digite o nome do produto: ")
preco_prod = float(input("Digite o preço do produto (R$): "))
categoria_prod = input("Digite a categoria do produto: ")

# Criando o dicionário base
produto = {
    "nome": nome_prod,
    "preco": preco_prod,
    "categoria": categoria_prod
}

# 3 e 4. Recebendo tags e removendo duplicatas com Set
tags_raw = input("Digite tags para o produto (separadas por vírgula): ")

# Criando lista sem espaços e convertendo para set
lista_tags = [tag.strip() for tag in tags_raw.split(",")]
tags_unicas = set(lista_tags)

# 5. Adicionando ao dicionário e exibindo
produto["tags"] = tags_unicas

print("\n📦 Cadastro do Produto Concluído:")
print(f"Nome: {produto['nome']}")
print(f"Preço: R$ {produto['preco']:.2f}")
print(f"Categoria: {produto['categoria']}")
print(f"Tags Únicas: {produto['tags']}")
```
