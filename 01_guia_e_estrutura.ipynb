## 🎯 Objetivo do Minicurso

O foco do minicurso é construir uma base sólida em lógica de programação e na sintaxe fundamental do Python. Ao concluir os 10 módulos, você dominará desde conceitos básicos como variáveis e operadores até estruturas de controle, coleções de dados e funções para automatizar tarefas e resolver problemas reais.

---

## 🗺️ Estrutura do Curso

A trilha de aprendizado é composta por **10 módulos práticos** desenvolvidos para uma evolução gradual:

| Módulo | Arquivo Notebook | Conteúdo Principal |
| :---: | :--- | :--- |
| **01** | `01_guia_e_estrutura.ipynb` | Guia e estrutura do curso |
| **02** | `02_introducao_ao_python.ipynb` | O que é Python, ecossistema Jupyter, função `print()` e boas práticas com comentários |
| **03** | `03_variaveis_e_tipos_de_dados.ipynb` | Conceito de variáveis, tipos primitivos (`int`, `float`, `str`, `bool`) e conversão de tipos (*casting*) |
| **04** | `04_entrada_de_dados_e_operadores.ipynb` | Interatividade com `input()`, operadores aritméticos e ordem de precedência |
| **05** | `05_manipulacao_de_strings.ipynb` | Formatação moderna com f-strings, métodos nativos para tratamento de texto e fatiamento (*slicing*) |
| **05** | `06_estruturas_condicionais.ipynb` | Lógica booleana, regras de identação no Python, tomadas de decisão com `if`, `elif` e `else` |
| **06** | `07_listas_e_tuplas.ipynb` | Estruturas sequenciais, métodos de manipulação de listas e o conceito de imutabilidade nas tuplas |
| **07** | `08_dicionarios_e_conjuntos.ipynb` | Mapeamento chave-valor em dicionários (`dict`) e operações de conjuntos únicos (`set`) |
| **08** | `09_estruturas_de_repeticao.ipynb` | Automação e iterações com os laços `for` e `while`, uso de `range()`, `break` e `continue` |
| **10** | `10_funcoes_e_modularizacao.ipynb` | Criação de funções reutilizáveis com `def`, envio de parâmetros, retorno com `return` e escopo de variáveis |
| **11** | `11_projeto_pratico_desafio.ipynb` | Projeto prático integrador para consolidar todo o conteúdo do minicurso com resolução comentada |
