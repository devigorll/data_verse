### 1. O que é Python?

O **Python** é uma linguagem de programação de **alto nível**, **interpretada**, de **propósito geral** e com uma sintaxe extremamente **simples e intuitiva**.

* **Alto Nível:** A escrita é muito próxima da linguagem humana (inglês), o que facilita a leitura e o aprendizado.
* **Interpretada:** O código é lido e executado linha por linha, sem a necessidade de um processo complexo de compilação prévia.
* **Multiparadigma:** Suporta programação orientada a objetos, funcional e estruturada.

---
